# RAG Intro — In-Memory Pipeline

This notebook demonstrates a minimal Retrieval-Augmented Generation (RAG) pipeline using an **in-memory** search index backed by [minsearch](https://github.com/alexeygrigorev/minsearch).

Pipeline steps:
1. **Load** FAQ documents from the LLM Zoomcamp dataset via HTTP (`FaqHttpLoader`)
2. **Index** the documents in memory (`MinsearchIndex`)
3. **Query** the pipeline with a natural-language question (`RAGBase`)
4. **Answer** is generated by the OpenAI API (`OpenAIClient`)

> **Note:** Requires `OPENAI_API_KEY` to be set in your environment or a `.env` file in the project root.

```python
import sys
sys.path.insert(0, '..')

from src import FaqHttpLoader, MinsearchIndex, RAGBase, OpenAIClient
```

```python
loader = FaqHttpLoader()
docs = loader.load()
print(f"Loaded {len(docs)} documents")
```

```python
index = MinsearchIndex(docs)
```

```python
rag = RAGBase(index=index, llm=OpenAIClient(), course_filter="data-engineering-zoomcamp")
```

```python
answer = rag.rag("How do I join the course?")
print(answer)
```
